# Import Datasets

```python
# load both datasets
bites = pd.read_csv("BABI_4005_P1/scripts/bites_cleaned.csv")
demographics = pd.read_csv("louisville_zip_demographics.csv")
```

# Convert Data Type to Object
- In order to merge the data sets together, the join keys must have the same data type
- Because victim_zip is type object, I will convert zip_code to type object

```python
demographics['zip_code'] = demographics['zip_code'].astype(str)
```

```python
# check
demographics.dtypes
```

# Join Datasets

```python
# join datasets using inner join
merged = pd.merge(bites, demographics, left_on='victim_zip', right_on='zip_code', how='inner')
merged
```

## Inspect New Merged Dataset

```python
merged.shape
```

## Explanation
1. I chose to use an inner join because in a business context, we'd want to analyze how ZIP code demographics influence bite incidents, and this type of analysis would require matching each bite record to its demographic record. Bite records without ZIP codes couldn't be used since there's no demographic data available, making it meaningless. Inner join keeps records where both bite location and area characteristics exist.
3. No new NA values were created because inner join naturally excludes non-matching records.
